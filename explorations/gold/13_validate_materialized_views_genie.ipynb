## Validate Genie's Answers
1. Which country has the most race result records overall?

```python
%sql
-- Validate Genie question:
-- Which country has the most race result records overall?

SELECT
    country_name,
    race_result_records
FROM marathos_catalog.gold.vw_global_leaderboard
ORDER BY race_result_records DESC
LIMIT 1;
```

```python
%sql
-- Manual validation from Gold fact table and country dimension

SELECT
    c.country_name,
    COUNT(*) AS race_result_records
FROM marathos_catalog.gold.fact_results f
INNER JOIN marathos_catalog.gold.dim_country c
    ON f.country_code_iso3 = c.country_code_iso3
GROUP BY
    c.country_name
ORDER BY
    race_result_records DESC
LIMIT 1;
```

2. What is the year that Sweden had the most marathon events held?

```python
%sql
-- Validate Genie question:
-- What year did Sweden have the most marathon events held?
--
-- Important:
-- In this dataset, country_name means athlete country.
-- So this validates the year where Swedish athletes appeared in the most unique events.

SELECT
    race_year,
    country_name,
    SUM(events_held) AS total_events_represented,
    SUM(athlete_entries) AS total_athlete_entries,
    SUM(unique_athletes) AS total_unique_athletes
FROM marathos_catalog.gold.vw_country_seasonal_trends
WHERE country_name = 'Sweden'
GROUP BY
    race_year,
    country_name
ORDER BY
    total_events_represented DESC
LIMIT 1;
```
