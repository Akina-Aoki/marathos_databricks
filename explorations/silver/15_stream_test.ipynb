```python
%sql
-- See that the fake data reached all layers
SELECT
    event_name,
    event_start_date,
    event_distance_length,
    athlete_country,
    country_name,
    athlete_gender,
    COUNT(*) AS row_count
FROM marathos_catalog.silver.marathon_results_obt
WHERE event_name IN (
    'Stockholm Marathon Innovation Run',
    'Paris Ultra City Challenge',
    'Tokyo Future Marathon'
)
GROUP BY
    event_name,
    event_start_date,
    event_distance_length,
    athlete_country,
    country_name,
    athlete_gender
ORDER BY
    event_name,
    athlete_gender;
```
